```python
import pandas as pd
from pandas_profiling import ProfileReport
import numpy as np
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
import random
```

```python
seed=39
random.seed(seed)
np.random.seed(seed)

X_train=pd.read_csv("DataSet/Train/X_train.txt",sep=" ",header=None)
y_train=pd.read_csv("DataSet/Train/y_train.txt",sep=" ",header=None)
X_test=pd.read_csv("DataSet/Test/X_test.txt",sep=" ",header=None)
y_test=pd.read_csv("DataSet/Test/y_test.txt",sep=" ",header=None)

# 70% training set, 30% orginal test set is furthur split into 15% final test set and 15% validation set
def split_validation_set(X,y,validation_size=0.5,random_state=39):
    X_train,X_val,y_train,y_val=train_test_split(X,y,test_size=validation_size,stratify=y,random_state=random_state)
    return X_train, X_val, y_train, y_val

X_final_test,X_final_val,y_final_test,y_final_val=split_validation_set(X_test,y_test)

print(y_train.shape)  
print(y_train.head())  
print(y_final_test.shape)
print(y_final_test.head())
print(y_final_val.shape)
print(y_final_val.head())
```

```python

target_classes=[7,8,9,10,11,12]
true_indices=y_train[y_train[0].isin(target_classes)].index
X_subset=X_train.loc[true_indices]
y_subset=y_train.loc[true_indices]


sampling_strategy_dict={}
for c in target_classes:
    sampling_strategy_dict[c]=900
sm=SMOTE(sampling_strategy=sampling_strategy_dict,random_state=39)
X_res,y_res=sm.fit_resample(X_subset,y_subset)

X_final_train=pd.concat([pd.DataFrame(X_res),pd.DataFrame(X_train)],ignore_index=True)
y_final_train=pd.concat([pd.DataFrame(y_res),pd.DataFrame(y_train)],ignore_index=True)

print(y_final_train.value_counts())

shuffled_indices = X_final_train.sample(frac=1, random_state=369).index
X_final_train = X_final_train.loc[shuffled_indices].reset_index(drop=True)
y_final_train = y_final_train.loc[shuffled_indices].reset_index(drop=True)

X_final_train.to_csv('DataSet/X_final_train.csv',index=False)
y_final_train.to_csv('DataSet/y_final_train.csv',index=False)
X_final_test.to_csv('DataSet/X_final_test.csv',index=False)
y_final_test.to_csv('DataSet/y_final_test.csv',index=False)
X_final_val.to_csv('DataSet/X_final_val.csv',index=False)
y_final_val.to_csv('DataSet/y_final_val.csv',index=False)


```
