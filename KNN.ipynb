```python
import numpy as np
import pandas as pd
import math
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from imblearn.over_sampling import SMOTE
from collections import Counter
from sklearn.model_selection import GridSearchCV

# Load data using NumPy (skip empty strings with genfromtxt)
X_train = np.genfromtxt('Train/X_train.txt')
y_train = np.genfromtxt('Train/y_train.txt')
X_test = np.genfromtxt('Test/X_test.txt')
y_test = np.genfromtxt('Test/y_test.txt')

# Print number of training and testing samples
print("Number of training samples:", X_train.shape[0])
print("Number of testing samples:", X_test.shape[0])

# Suggested value for k
print("Suggested k (floor of sqrt(7767)):", math.floor(math.sqrt(7767)))

# Value counts for training labels
unique, counts = np.unique(y_train, return_counts=True)
print("Value counts for training labels:")
for label, count in zip(unique, counts):
    print(f"Label {int(label)}: {count} samples")
```

```python
# Apply SMOTE
smote = SMOTE(sampling_strategy="auto",random_state=42)
X_train, y_train = smote.fit_resample(X_train, y_train)

# Check class distribution after SMOTE
print("\nAfter SMOTE:")
resampled_counts = Counter(y_train)
for label in sorted(resampled_counts):
    print(f"Label {label}: {resampled_counts[label]} samples")


print("Number of training samples:", X_train.shape[0])

print("Suggested k (floor of sqrt(training size)):", math.floor(math.sqrt(X_train.shape[0])))
```

```python
# Initialize KNN model
knn = KNeighborsClassifier(n_neighbors=5)

# Train model
knn.fit(X_train, y_train)
```

```python
# Predict
y_pred = knn.predict(X_test)

# Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred))
#print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```

2. PCA to reduce dimension

```python
from sklearn.decomposition import PCA

# PCA with 98% variance
pca1 = PCA(n_components=0.98)
X_train_pca1 = pca1.fit_transform(X_train)
X_test_pca1 = pca1.transform(X_test)
print("Number of components chosen for 98% variance:", pca1.n_components_)

# PCA with 95% variance
pca2 = PCA(n_components=0.95)
X_train_pca2 = pca2.fit_transform(X_train)
X_test_pca2 = pca2.transform(X_test)
print("Number of components chosen for 95% variance:", pca2.n_components_)

# PCA with 90% variance
pca3 = PCA(n_components=0.90)
X_train_pca3 = pca3.fit_transform(X_train)
X_test_pca3 = pca3.transform(X_test)
print("Number of components chosen for 90% variance:", pca3.n_components_)

#PCA with 99% variance
pca4 = PCA(n_components=0.99)
X_train_pca4 = pca4.fit_transform(X_train)
X_test_pca4 = pca4.transform(X_test)
print("Number of components chosen for 99% variance:", pca4.n_components_)
```

```python
#use pipeline to test the accuracy
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_val_score

for n in [38, 69, 116, 155, 561]:
    pipeline = Pipeline([
        ('pca', PCA(n_components=n)),
        ('knn', KNeighborsClassifier(n_neighbors=6))
    ])
    score = cross_val_score(pipeline, X_train, y_train, cv=5, scoring='accuracy')
    print(f"PCA components: {n}, Accuracy: {score.mean():.4f}")
```
